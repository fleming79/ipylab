```python
from ipylab import SimpleOutput
from ipylab.widgets import KeyboardCapture
from ipywidgets import VBox
```

```python
kc = KeyboardCapture(description="Select to capture keystrokes")
```

```python
out = SimpleOutput(max_outputs=4)
kc.register(out.push)
VBox([out, kc])
```

```python
out.max_outputs = 3
```
